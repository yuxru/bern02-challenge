## Model Writeup

**Project question**: given only a molecule's 2D structure, how accurately can we predict its B3LYP gap for molecules the model has never seen?

**Gradient Boosting Regressor**:
1. Compute residuals. rᵢ = yᵢ − current prediction = how far off each molecule's gap still is (eV).

2. Fit regression tree to the residuals which splits the molecules on questions like "n_conjugated_bonds <= 3?". At each split picks variable and cut point that reduce Residual Sum Squares most.

- Ridge and lasso shrink coefficients towards 0 so no single predictor dominates. Boosting shrinks each tree so no single tree dominates.

- Each feature is binned into at most 255 quantile bins before fitting: each split checks about 255 cut points instead of about 100,000 unique values.
